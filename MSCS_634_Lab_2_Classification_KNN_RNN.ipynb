## Step 1: Load and Prepare the Dataset

This notebook loads the Wine dataset from `sklearn`, explores the features and target distribution, and splits the data into training and testing sets.

```python
import pandas as pd
import numpy as np
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier, RadiusNeighborsClassifier
from sklearn.metrics import accuracy_score
import matplotlib.pyplot as plt

plt.style.use('seaborn-v0_8')

wine = load_wine()
X = pd.DataFrame(wine.data, columns=wine.feature_names)
y = pd.Series(wine.target, name='target')

print('Feature shape:', X.shape)
print('Target shape:', y.shape)
```

```python
display(X.head())
print('
Class distribution:')
print(y.value_counts().sort_index())
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

print('Training set size:', X_train.shape[0])
print('Test set size:', X_test.shape[0])
```

## Step 2: Implement K-Nearest Neighbors (KNN)

Train and evaluate KNN models for multiple values of `k`.

```python
knn_values = [1, 5, 11, 15, 21]
knn_accuracies = []

for k in knn_values:
    model = KNeighborsClassifier(n_neighbors=k)
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    acc = accuracy_score(y_test, y_pred)
    knn_accuracies.append(acc)
    print(f'KNN k={k}: accuracy={acc:.4f}')
```

## Step 3: Implement Radius Neighbors (RNN)

Train and evaluate RNN models for several radius values and record accuracy.

```python
rnn_values = [350, 400, 450, 500, 550, 600]
rnn_accuracies = []

for r in rnn_values:
    model = RadiusNeighborsClassifier(radius=r)
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    acc = accuracy_score(y_test, y_pred)
    rnn_accuracies.append(acc)
    print(f'RNN radius={r}: accuracy={acc:.4f}')
```

## Step 4: Visualize and Compare Results

Plot accuracy trends for KNN and RNN, then compare model performance.

```python
fig, ax = plt.subplots(1, 2, figsize=(14, 5))

ax[0].plot(knn_values, knn_accuracies, marker='o', color='tab:blue')
ax[0].set_title('KNN Accuracy vs k')
ax[0].set_xlabel('k (n_neighbors)')
ax[0].set_ylabel('Accuracy')
ax[0].set_xticks(knn_values)
ax[0].grid(True, linestyle='--', alpha=0.5)

ax[1].plot(rnn_values, rnn_accuracies, marker='o', color='tab:green')
ax[1].set_title('RNN Accuracy vs Radius')
ax[1].set_xlabel('Radius')
ax[1].set_ylabel('Accuracy')
ax[1].set_xticks(rnn_values)
ax[1].grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

## Observations and Comparison

- KNN typically performs well with a moderate number of neighbors, balancing bias and variance.
- RNN with large radius values tends to include more neighbors and may default to the majority class if all points are inside the radius.
- In this dataset, KNN may be preferable when the decision boundary is influenced by a small neighborhood, while RNN can be useful if distance-based grouping at a specific scale is desired.
